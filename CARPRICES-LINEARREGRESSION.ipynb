```python
price = pd.read_csv("CarPrice_Assignment.csv")
price = price.drop(columns="car_ID")
price["CarName"] = price["CarName"].str.split().str[0].str.lower()
```

```python
num_cols = price.select_dtypes(include=np.number).columns.drop('price')

corr_df = pd.DataFrame({
    'Pearson (price)':      price[num_cols].corrwith(price['price']),
    'Spearman (log price)': price[num_cols].corrwith(np.log(price['price']), method='spearman'),
}).sort_values('Spearman (log price)')

corr_df.plot.barh(figsize=(8, 7), title='Correlation of numeric features with price')
plt.axvline(0, color='black', lw=0.8)
plt.tight_layout()
plt.show()
```

```python
data = pd.get_dummies(price, dtype=float)
```

```python
y = np.log(data["price"].values)
x=data.drop(columns="price").values.astype(float)
```

```python
split=int(0.8*len(x))
idx=np.arange(len(x))
rng=np.random.default_rng(42)
rng.shuffle(idx)
train_idx,test_idx=idx[:split],idx[split:]
x_train,x_test=x[train_idx],x[test_idx]
y_train,y_test=y[train_idx],y[test_idx]
```

```python
X = x_train.mean(axis=0)
sigma= x_train.std(axis=0)
sigma[sigma == 0] = 1
x_train = (x_train - X) / sigma
x_test = (x_test - X) / sigma
```

```python
x_train = np.c_[np.ones(len(x_train)), x_train]
x_test = np.c_[np.ones(len(x_test)), x_test]
print(x_train.shape, x_test.shape)
```

```python
def gradient_descent(x, y, alpha=0.05, iters=200,lam=0.1):
    m, n = x.shape
    w = np.zeros(n)
    costs = []
    for _ in range(iters):
        err = x @ w - y
        reg = lam * w; reg[0] = 0
        w -= alpha * (x.T @ err + reg) / m
        costs.append((err @ err) / (2 * m))
    return w, costs

w, costs = gradient_descent(x_train, y_train)
```

```python
lr_hist = {}
for alpha in [0.001, 0.01, 0.05, 0.1]:
    _, h = gradient_descent(x_train, y_train, alpha=alpha, iters=300,lam=10 )
    lr_hist[f'alpha={alpha}'] = h
pd.DataFrame(lr_hist).plot(figsize=(7,4), title='Effect of learning rate on training loss', logy=True,
                           xlabel='Epoch', ylabel='Loss (log scale)')
```

```python
lam_hist = {}
for lam in [0, 0.1, 1, 10, 30, 100]:
    _, h = gradient_descent(x_train, y_train, alpha=0.1, iters=300, lam=lam)
    lam_hist[f'lam={lam}'] = h

pd.DataFrame(lam_hist).plot(figsize=(7, 4), title='Effect of lam on training loss', logy=True,
                            xlabel='Epoch', ylabel='Loss (log scale)')
```

```python
plt.plot(costs)
plt.xlabel("Epochs"); plt.ylabel("Loss"); plt.title("Loss vs epochs")
plt.show()
```

```python
def rmse(y_true, y_pred):
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

def r2(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - y_true.mean()) ** 2)
    return 1 - ss_res / ss_tot

def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

pred_train = x_train @ w
pred_test = x_test @ w
print("Train RMSE:", rmse(y_train, pred_train), " R2:", r2(y_train, pred_train),"Train  MAE:", mae(y_train, pred_train))
print("Test  RMSE:", rmse(y_test, pred_test), " R2:", r2(y_test, pred_test),"Test  MAE:", mae(y_test, pred_test))
```
